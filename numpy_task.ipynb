```python
import numpy as np

zeros = np.zeros((3, 3))
ones = np.ones((2, 4))
identity = np.eye(4)

print("Zeros Matrix:")
print(zeros)

print("Ones Matrix:")
print(ones)

print("Identity Matrix:")
print(identity)
```

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

result = A @ B
transpose = A.T

print("Matrix Multiplication:")
print(result)

print("Transpose of A:")
print(transpose)
```

```python
import numpy as np

marks = np.array([78, 85, 90, 62, 88])

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Standard Deviation:", np.std(marks))
```

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

combined = np.concatenate((a, b))
values = np.linspace(0, 10, 5)

print("Concatenated Array:", combined)
print("Linspace Values:", values)
```

```python
import numpy as np

arr = np.array([23, 45, 12, 67, 34])

print("Argmax Index:", np.argmax(arr))
print("Argmin Index:", np.argmin(arr))

result = np.where(arr > 30, arr, 0)

print("Values Greater Than 30:", arr[arr > 30])
```

```python
import numpy as np

# Task 10 - Sales Analysis Project

# Part 1 - Cumulative Sum
sales = np.array([100, 200, 150, 300])

cumulative = np.cumsum(sales)

print("Cumulative Sum:", cumulative)


# Part 2 - Correlation Coefficient
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([20, 40, 50, 70, 90])

correlation = np.corrcoef(hours, marks)

print("Correlation Coefficient:", correlation[0, 1])


# Part 3 - Save and Load
arr = np.array([10, 20, 30, 40])

np.save("numbers.npy", arr)

loaded = np.load("numbers.npy")

print("Loaded Array:", loaded)
```
