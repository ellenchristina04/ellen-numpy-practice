```python
import re

text = "Call me at 9876543210 or 9123456780"

result = re.findall(r"\d{10}", text)

print(result)
```

```python
import re

text = "Emails: john@gmail.com, sara@yahoo.com"

result = re.findall(r"\w+@\w+\.\w+", text)

print(result)
```

```python
import re

text = "Order ID: 45678"

result = re.search(r"\d+", text)

print(result.group())
```

```python
import re

text = "EMP12345 joined today"

result = re.match(r"^EMP", text)

if result:
    print("Match Found")
else:
    print("No Match")
```

```python
import re

text = "ABCDE1234F"

result = re.fullmatch(r"[A-Z]{5}\d{4}[A-Z]", text)

if result:
    print("Valid PAN")
else:
    print("Invalid PAN")
```
