```python
#1. Heights mean=170, SD=10; probability >180.

# Import the normal distribution function
from scipy.stats import norm

# Define the mean height
mean = 170

# Define the standard deviation
sd = 10

# Define the height value
x = 180

# Calculate the probability of height greater than 180
probability = 1 - norm.cdf(x, mean, sd)

# Display the probability
print("Probability =", probability)
```
