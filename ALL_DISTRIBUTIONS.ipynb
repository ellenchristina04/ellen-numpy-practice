```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# Select data
data = df['RATING'].dropna()


# 1. Normal Q-Q Plot
plt.figure(figsize=(6, 5))

stats.probplot(data, dist="norm", plot=plt)

plt.title("Normal Q-Q Plot - Movies Score")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Ordered Math Scores")

plt.show()
```

```python
plt.figure(figsize=(6, 5))

stats.probplot(data, dist=stats.expon, plot=plt)

plt.title("Exponential Q-Q Plot - Math Score")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Ordered Math Scores")

plt.show()
```

```python
 # 3. Uniform Q-Q Plot
plt.figure(figsize=(6, 5))

stats.probplot(data, dist=stats.uniform, plot=plt)

plt.title("Uniform Q-Q Plot - Math Score")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Ordered Math Scores")

plt.show()
```

```python
 # 4. t-Distribution Q-Q Plot
plt.figure(figsize=(6, 5))

stats.probplot(
    data,
    dist=stats.t,
    sparams=(5,),
    plot=plt
)

plt.title("t-Distribution Q-Q Plot - Math Score")
plt.xlabel("Theoretical Quantiles")
plt.ylabel("Ordered Math Scores")

plt.show()
```
