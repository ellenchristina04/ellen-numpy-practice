```python
# to find outliers on numerical terms
movie_columns = ['RATING', 'RunTime']

df[movie_columns].head()
```

```python
# let create boxplot for movie columns
plt.figure(figsize=(12,6))

sns.boxplot(data=df[movie_columns])

plt.title("Boxplot for Movie Data")

plt.xticks(rotation=45)

plt.show()
```

```python
rating = df['RATING']

Q1 = rating.quantile(0.25)
Q3 = rating.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1", Q1)
print("Q3", Q3)
print("IQR", IQR)
print("lower_bound", lower_bound)
print("upper_bound", upper_bound)
```

```python
for col in movie_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
outlier_summary = []

for col in movie_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append([
        col,
        Q1,
        Q3,
        IQR,
        lower,
        upper,
        count
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=[
        'Column',
        'Q1',
        'Q3',
        'IQR',
        'Lower_Bound',
        'Upper_Bound',
        'Outlier_Count'
    ]
)

print(outlier_df)
```

```python
df_clean = df.copy()
```

```python
for col in movie_columns:

    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_clean = df_clean[
        (df_clean[col] >= lower) &
        (df_clean[col] <= upper)
    ]
```

```python
sns.histplot(df_clean['RATING'], kde=True)

plt.title("Rating Distribution After Outlier Removal")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")

plt.show()
```
