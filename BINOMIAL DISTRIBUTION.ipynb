```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import stats
from scipy.stats import kurtosis
```

```python
# Task 1: Defect-Free Items

from scipy.stats import binom

n = 10
p = 0.95
x = 9

# Calculate probability
probability = binom.pmf(x, n, p)

print("Probability of exactly 9 good items:", round(probability, 4))
```

```python
# Task 2: Student Solves Problems

from scipy.stats import binom

n = 5
p = 0.60

# Calculate probability of at least 3 successes
probability = binom.sf(2, n, p)

print("Probability of solving at least 3 problems:", round(probability, 4))
```

```python
# Task 3: Basketball Free-Throw Success

from scipy.stats import binom

n = 8
p = 0.70
x = 6

# Calculate probability
probability = binom.pmf(x, n, p)

print("Probability of making exactly 6 shots:", round(probability, 4))
```

```python
# Task 4: Quiz Guessing

from scipy.stats import binom

n = 10
p = 0.25
x = 4

# Calculate probability
probability = binom.pmf(x, n, p)

print("Probability of exactly 4 correct answers:", round(probability, 4))
```
