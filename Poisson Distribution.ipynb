```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import stats
from scipy.stats import kurtosis
```

```python
#Task 1: Website gets 6 visits/minute; probability of exactly 10 visits next minute
from scipy.stats import poisson

rate = 6
visits = 10

probability = poisson.pmf(visits, rate)

print("Probability:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
```

```python
#Task 2: Bank gets 3 customers/5 minutes; probability of 5 customers in next 5 minutes
from scipy.stats import poisson

average = 3
customers = 5

probability = poisson.pmf(customers, average)

print("Probability:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
```

```python
#Task 3: Hospital gets 10 patients/hour; probability of more than 12 patients next hour
from scipy.stats import poisson

rate = 10
patients = 12

probability = poisson.sf(patients, rate)

print("Probability:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
```

```python
#Task 4: Call center gets 12 calls/hour; probability of no calls in 5 minutes
from scipy.stats import poisson

rate = 12 / 12
calls = 0

probability = poisson.pmf(calls, rate)

print("Probability:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
```

```python
#Task 5: Toll booth sees 4 cars/minute; probability of 2 cars in 30 seconds
from scipy.stats import poisson

rate = 4 * 0.5
cars = 2

probability = poisson.pmf(cars, rate)

print("Probability:", round(probability, 4))
print("Percentage:", round(probability * 100, 2), "%")
```
