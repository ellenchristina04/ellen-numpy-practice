```python
#1. Customer arrives every 10 min; probability within 3 min.

# Import the exponential distribution function
from scipy.stats import expon

# Define the average arrival time
mean_time = 10

# Define the time limit
x = 3

# Calculate the probability of arrival within 3 minutes
probability = expon.cdf(x, scale=mean_time)

# Display the probability
print("Probability =", probability)
```

```python
#2. System crash avg 5 hrs; probability crash within 1 hr.

# Import the exponential distribution function
from scipy.stats import expon

# Define the average time between crashes
mean_time = 5

# Define the time limit
x = 1

# Calculate the probability of crash within 1 hour
probability = expon.cdf(x, scale=mean_time)

# Display the probability
print("Probability =", probability)
```
