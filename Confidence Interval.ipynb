```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import stats
from scipy.stats import kurtosis
```

```python
#1. 50 students, mean=165, S2. Bottle fill mean=500ml, SD=8, n=40; 99% CI.D=10; 95% CI.

from scipy.stats import norm
import math

mean = 165
sd = 10
n = 50
confidence = 0.95

z = norm.ppf((1 + confidence) / 2)
margin = z * sd / math.sqrt(n)

lower = mean - margin
upper = mean + margin

print("95% Confidence Interval:", round(lower, 2), "to", round(upper, 2))
```
