```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

data = {
    "Age": [25,30,35,40,45],
    "Salary": [25000,35000,45000,55000,65000],
    "Experience": [1,2,3,4,5]
}

df = pd.DataFrame(data)

correlation = df.corr()

sns.heatmap(correlation, annot=True)
plt.title("Correlation Matrix")
plt.show()
```
