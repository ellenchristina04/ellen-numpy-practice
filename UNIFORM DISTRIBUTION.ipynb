```python
#1. Lucky draw 1–100; probability 30–60.

# Import the uniform distribution function
from scipy.stats import uniform

# Define the minimum value
minimum = 1

# Define the maximum value
maximum = 100

# Define the lower limit
lower = 30

# Define the upper limit
upper = 60

# Calculate the probability between 30 and 60
probability = (upper - lower) / (maximum - minimum)

# Display the probability
print("Probability =", probability)
```

```python
#2. Bus arrival 0–20 min; probability <5 min.

# Import the uniform distribution function
from scipy.stats import uniform

# Define the minimum arrival time
minimum = 0

# Define the maximum arrival time
maximum = 20

# Define the time limit
x = 5

# Calculate the probability of bus arrival before 5 minutes
probability = (x - minimum) / (maximum - minimum)

# Display the probability
print("Probability =", probability)
```
