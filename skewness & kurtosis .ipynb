```python
import matplotlib.pyplot as plt
import pandas as pd
from scipy.stats import stats
from scipy.stats import kurtosis
```

```python
# Task 1: E-Commerce Order Value Analysis

orders = pd.Series([200, 250, 300, 350, 400, 450, 5000])

# Draw Histogram
orders.plot(kind="hist", bins=6, edgecolor="black")
plt.title("E-Commerce Order Value")
plt.xlabel("Order Value")
plt.ylabel("Frequency")
plt.show()

# Calculate Skewness
value = orders.skew()

# Identify the type
if value > 0:
    result = "Positive Skew"
elif value < 0:
    result = "Negative Skew"
else:
    result = "Symmetric Distribution"

print("Skewness Value:", round(value, 4))
print("Distribution Type:", result)

# Final Conclusion
print("Conclusion: Few customers place huge orders.")
```
