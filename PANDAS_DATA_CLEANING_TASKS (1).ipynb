```python
import pandas as pd
import numpy as np

data = {
    "Student": ["Asha", "Bala", "Charan", "Divya", "Eshan", "Farah"],
    "Maths": [90, np.nan, 78, 85, np.nan, 67],
    "Science": [88, 92, np.nan, 80, 75, np.nan],
    "English": [76, 81, np.nan, 79, 85, 88],
    "Attendance": [95, np.nan, 89, 91, 87, np.nan],
    "Section": ["A", "B", "A", np.nan, "B", "A"],
    "City": ["Chennai", "Salem", np.nan, "Madurai", "Salem", "Chennai"]
}

df = pd.DataFrame(data)

# Fill marks using mean
df["Maths"] = df["Maths"].fillna(df["Maths"].mean())
df["Science"] = df["Science"].fillna(df["Science"].mean())
df["English"] = df["English"].fillna(df["English"].mean())

# Fill Attendance using median
df["Attendance"] = df["Attendance"].fillna(df["Attendance"].median())

# Fill Section and City using mode
df["Section"] = df["Section"].fillna(df["Section"].mode()[0])
df["City"] = df["City"].fillna(df["City"].mode()[0])

print(df)
```

```python
import pandas as pd
import numpy as np

data = {
    "Patient": ["P1", "P2", "P3", "P4", "P5", "P6"],
    "Age": [45, np.nan, 50, 39, 60, np.nan],
    "Blood_Pressure": [120, 130, np.nan, 110, np.nan, 140],
    "Sugar_Level": [150, np.nan, 170, 120, 180, 160],
    "Weight": [72, 80, np.nan, 65, 78, np.nan],
    "Disease": ["Diabetes", np.nan, "Fever", "Fever", "Diabetes", "Cold"],
    "Doctor": ["Dr.A", "Dr.B", np.nan, "Dr.A", "Dr.C", "Dr.B"]
}

df = pd.DataFrame(data)

# Numerical columns
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Blood_Pressure"] = df["Blood_Pressure"].fillna(df["Blood_Pressure"].mean())
df["Sugar_Level"] = df["Sugar_Level"].fillna(df["Sugar_Level"].mean())
df["Weight"] = df["Weight"].fillna(df["Weight"].median())

# Categorical columns
df["Disease"] = df["Disease"].fillna(df["Disease"].mode()[0])
df["Doctor"] = df["Doctor"].fillna(df["Doctor"].mode()[0])

print(df)
```

```python
import pandas as pd
import numpy as np

data = {
    "Movie": ["Leo", "Master", "Vikram", "Jailer", "Beast", "Kaithi"],
    "Rating": [8.5, np.nan, 9.1, 8.9, np.nan, 8.7],
    "Votes": [1200, 980, np.nan, 1500, 1100, np.nan],
    "Collection_Cr": [300, np.nan, 450, 500, 250, 320],
    "Duration_Min": [145, 160, np.nan, 170, 155, 148],
    "Language": ["Tamil", "Tamil", np.nan, "Tamil", "Tamil", "Tamil"],
    "Platform": ["Netflix", np.nan, "Hotstar", "Amazon", "Netflix", "Hotstar"]
}

df = pd.DataFrame(data)

# Mean
df["Rating"] = df["Rating"].fillna(df["Rating"].mean())
df["Collection_Cr"] = df["Collection_Cr"].fillna(df["Collection_Cr"].mean())

# Median
df["Votes"] = df["Votes"].fillna(df["Votes"].median())
df["Duration_Min"] = df["Duration_Min"].fillna(df["Duration_Min"].median())

# Mode
df["Language"] = df["Language"].fillna(df["Language"].mode()[0])
df["Platform"] = df["Platform"].fillna(df["Platform"].mode()[0])

print(df)
```
