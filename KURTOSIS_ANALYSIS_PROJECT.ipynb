```python
#Import Libraries

import numpy as np              # Used for mathematical calculations
import pandas as pd             # Used to read and handle the dataset
import matplotlib.pyplot as plt # Used to create graphs
import seaborn as sns           # Used to create statistical graphs
from scipy.stats import kurtosis as k  # Used to calculate kurtosis
```

```python
#Load Dataset

df = pd.read_csv("/content/student_exam_performance.csv")
 # Used to load the CSV dataset

print(df.head())
 # Used to display the first five rows
```

```python
 #Select Column

 col_name = "exam_score"  # Used to select the exam_score column for kurtosis analysis
```

```python
#Remove Missing Values

data = df[col_name].dropna()  # Used to remove missing values from the selected column
```

```python
#Fisher Kurtosis

kurt_f = k(data, fisher=True)  # Used to calculate Fisher kurtosis

print("Fisher Kurtosis:", kurt_f)  # Used to display the Fisher kurtosis value
```

```python
#Pearson Kurtosis

kurt_p = k(data, fisher=False)  # Used to calculate Pearson kurtosis

print("Pearson Kurtosis:", kurt_p)  # Used to display the Pearson kurtosis value
```

```python
#Mean

mean_val = np.mean(data)  # Used to calculate the average exam score

print("Mean:", mean_val)  # Used to display the mean value
```

```python
#Minimum

min_val = np.min(data)  # Used to find the lowest exam score

print("Minimum:", min_val)  # Used to display the minimum value
```

```python
#Maximum

max_val = np.max(data)  # Used to find the highest exam score

print("Maximum:", max_val)  # Used to display the maximum value
```

```python
#Kurtosis Classification

if kurt_f > 0:  # Used to check whether Fisher kurtosis is positive
    print("Leptokurtic")

elif kurt_f < 0:  # Used to check whether Fisher kurtosis is negative
    print("Platykurtic")

else:  # Used when Fisher kurtosis is equal to zero
    print("Mesokurtic")
```

```python
#Histogram

plt.figure(figsize=(8, 4))  # Used to set the graph size

plt.hist(data, bins=30, edgecolor="black")  # Used to display the exam score distribution

plt.xlabel("Exam Score")  # Used to label the x-axis
plt.ylabel("Frequency")  # Used to label the y-axis
plt.title("Exam Score Distribution")  # Used to give a title to the graph

plt.show()  # Used to display the graph
```

```python
#Median

median_val = np.median(data)  # Used to calculate the middle exam score

print("Median:", median_val)  # Used to display the median value
```

```python
#Line Plot

plt.plot(data.values[:300])  # Used to plot the first 300 exam score observations

plt.axhline(mean_val, linestyle="--")  # Used to show the mean value as a reference line

plt.axhline(median_val, linestyle="--")  # Used to show the median value as a reference line

plt.show()  # Used to display the line plot
```

```python
#. Histogram + KDE

sns.histplot(data, kde=True, bins=30)  # Used to show the distribution and smooth KDE curve

plt.show()  # Used to display the graph
```

```python
#Mean and Median on Histogram

sns.histplot(data, kde=True, bins=30)  # Used to display the exam score distribution

plt.axvline(mean_val, linestyle="--")  # Used to mark the mean value on the distribution

plt.axvline(median_val, linestyle="--")  # Used to mark the median value on the distribution

plt.show()  # Used to display the graph
```
